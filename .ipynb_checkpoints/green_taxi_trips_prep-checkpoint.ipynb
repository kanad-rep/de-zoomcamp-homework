```python
# Defining the url variables
green_taxi_url = "https://d37ci6vzurychx.cloudfront.net/trip-data/green_tripdata_2025-11.parquet"
zones_url = "https://github.com/DataTalksClub/nyc-tlc-data/releases/download/misc/taxi_zone_lookup.csv"
```

```python
# Reading the trips data file
df_trips = pd.read_parquet(green_taxi_url)
df_trips.head()
```

```python
# Reading the zone data file
df_zones = pd.read_csv(zones_url)
df_zones.head()
```

```python
# Converting the necessary field to timestamp
df_trips['lpep_pickup_datetime'] = pd.to_datetime(df_trips['lpep_pickup_datetime'])
```

### Question 3

```python
# Selecting trip data in the month of Nov 2025
nov_trips = df_trips[(df_trips['lpep_pickup_datetime'] >= '2025-11-01') & (df_trips['lpep_pickup_datetime'] < '2025-12-01')].reset_index(drop = True)
nov_trips.head()
```

```python
# Displaying the min and max pick up timestamps
print('Min Date:', min(nov_trips.lpep_pickup_datetime))
print('Max Date:', max(nov_trips.lpep_pickup_datetime))
```

```python
# Displaying the columns
nov_trips.columns
```

```python
print('Trips in Nov 2025 with trip distance less than or equal to 1 mile:', nov_trips[nov_trips['trip_distance'] <= 1].shape[0])
```

### Question 4

```python
# Filtering the data for trips less than 100 miles
trips_100 = df_trips[df_trips['trip_distance'] < 100].reset_index(drop = True)

# Getting the row with the longest trip distance
longest_trip = max(trips_100['trip_distance'])

# Getting the date of the longest trip
pickup_date = trips_100[trips_100['trip_distance'] == longest_trip ]['lpep_pickup_datetime'].dt.date.iloc[0]

# Displaying the output
print('Pickup Day with the longest trip:', pickup_date)
```

### Question 5

```python
nov_trips
```

```python
# Merging the trips data with the zone data based on pickup location id
merged_df1 = nov_trips.merge(df_zones, how = 'left', left_on = 'PULocationID', right_on = 'LocationID')

# Filtering the data for data wrt to pickups on Nov 18, 2025
nov18_trips = merged_df1[(merged_df1['lpep_pickup_datetime'] >= "2025-11-18") & (merged_df1['lpep_pickup_datetime'] < "2025-11-19")]
nov18_trips.head()
```

```python
# Getting the zone with the highest amount
grouped_df1 = nov18_trips.groupby('Zone', as_index = False).agg({'total_amount' : 'sum'}).sort_values('total_amount', ascending = False, ignore_index = True)
grouped_df1.head()
```

```python
print('PickUp Zone with the highest amount on Nov 18, 2025:', grouped_df1.Zone[0])
```

### Question 6

```python
# Filtering the data for pick up location as East Harlem North
ehn_df = merged_df1[merged_df1['Zone'] == "East Harlem North"].reset_index(drop = True)

# Merging the zones data wrt dropoff location ids
merged_df2 = ehn_df.merge(df_zones, how = 'left', left_on = 'DOLocationID', right_on = 'LocationID')

# Renaming the conflicting columns for better understanding
merged_df2.rename(columns = {'Borough_x' : 'PUBorough', 'Zone_x' : 'PUZone', 'service_zone_x' : 'PU_service_zone', 
                             'Borough_y' : 'DOBorough', 'Zone_y' : 'DOZone', 'service_zone_y' : 'DO_service_zone'}, inplace = True)
merged_df2.head()
```

```python
merged_df2.columns
```

```python
# Getting the record with the largest tip amount
largest_tip = max(merged_df2['tip_amount'])

# Getting the dropoff zone with the largest tip amount
do_zone = merged_df2[merged_df2['tip_amount'] == largest_tip].DOZone.iloc[0]
print('DropOff Zone with the largest tip:', do_zone)
```
