```python
import pandas as pd
```

```python
url = "https://d37ci6vzurychx.cloudfront.net/trip-data/green_tripdata_2025-10.parquet"
```

```python
columns  = ["lpep_pickup_datetime","lpep_dropoff_datetime","PULocationID","DOLocationID","passenger_count",
"trip_distance","tip_amount","total_amount"]
df = pd.read_parquet(url, columns=columns)
df.head()
```

```python
df.shape
```

```python
from models import Ride, ride_from_dict, ride_serializer
```

```python
from kafka import KafkaProducer

server = "localhost:9092"
producer = KafkaProducer(bootstrap_servers=server, value_serializer=ride_serializer)
```

```python
import dataclasses

topic_name = 'green-trips'

# producer.send(topic_name, value=ride)
# producer.flush()
```

```python
import time

data_records = df.to_dict('records')


t0 = time.time()

counter = 1
for row_dict in data_records:
    ride = ride_from_dict(row_dict)
    producer.send(topic_name, value=ride)
    print(f"Sent: Row {counter}")
    time.sleep(0.01)
    counter += 1

producer.flush()

t1 = time.time()
print(f'took {(t1 - t0):.2f} seconds')
```
