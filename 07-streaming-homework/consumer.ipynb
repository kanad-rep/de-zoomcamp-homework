```python
from kafka import KafkaConsumer
import json

server = "localhost:9092"
topic_name = 'green-trips'
```

```python
from models import Ride, ride_deserializer
```

```python
consumer = KafkaConsumer(
    topic_name,
    bootstrap_servers=[server],
    auto_offset_reset='earliest',
    group_id='rides-console-v2',
    consumer_timeout_ms=5000,  # 👈 stops after idle
    value_deserializer=ride_deserializer
)
```

```python
count = 0
total = 0

print("Reading messages...")

for message in consumer:
    print("Message received")
    trip = message.value
    total += 1

    # Handle missing/null values safely
    distance = trip.trip_distance

    if distance is not None and float(distance) > 5.0:
        count += 1

    # Optional: stop condition (if dataset is finite)
    # remove this if you want continuous streaming
    if total % 10000 == 0:
        print(f"Processed {total} messages...")

    # If you know dataset size, you can break manually
    # Example:
    if total >= 50000:
        break

print(f"\nTotal trips: {total}")
print(f"Trips with distance > 5 km: {count}")
```
